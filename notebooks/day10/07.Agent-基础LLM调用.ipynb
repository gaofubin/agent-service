```python
from IPython.core.display import Image
from typing import TypedDict

from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
from langgraph.constants import START, END
from langgraph.graph import StateGraph

load_dotenv()

# 1.初始化模型
model = init_chat_model(
    "deepseek-v4-flash",
    extra_body={"thinking": {"type": "disabled"}}
)

# 2.创建State
class LLMState(TypedDict):
    input: str
    output: str

# 3.创建Node
def llm_node(state: LLMState):
    # 调用模型，传入str，返回 AIMessage
    response = model.invoke(state["input"])
    return {"output": response.text}

# 4.添加Edge，连接Node，形成Graph
builder = StateGraph(LLMState)

builder.add_node("llm", llm_node)

builder.add_edge(START, "llm")
builder.add_edge("llm", END)

graph = builder.compile()

display(Image(graph.get_graph(xray=True).draw_mermaid_png()))
```

调用model和调用Agent：
- 入参：
    - model, 传入的str或者`list[BaseMessage]`
    - agent, 传入的是dict，dict里有messages，是一个消息列表`list[BaseMessage]`
- 返回值：
    - model，返回的AIMessage
    - agent，返回的是dict，dict里有messages，是一个消息列表`list[BaseMessage]`

```python
# 5.调用Graph
result = graph.invoke({"input": "你好，你是谁？"})
print(result)
```
