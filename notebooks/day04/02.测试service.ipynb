```python
from app.infra.database import AsyncSessionFactory
from app.modules.chat_thread.models import ChatThread

async with AsyncSessionFactory() as session:
    async with session.begin():
        # 根据id查询
        chat_thread = await session.get(ChatThread, '625c91ae-59ce-45b7-8795-55f124c02932')
        print("old : ", chat_thread)

        # 修改字段值，就能触发更新
        chat_thread.title="测试14"

        # 手动刷新
        await session.flush()
        await session.refresh(chat_thread)

        print(chat_thread)
```
